#1

- convert the setence into words and lowercase
- create empty dictioanry to store them
- count the words 
- return dictioanry

```python
def freq(sentence):
    words = sentence.lower().split()
    s = {}
    for word in words:
        if word in s:
            s[word] +=1

        else:
            s[word] = 1
    return s
sentence = "python is easy and python is powerful"
r = freq(sentence)
print(r)
print()
sentence = "data analysis is data visualization, data cleaning and data processing"
e = freq(sentence)
print(e)
```

#4
- convert all strings into lowercase and remove spaces
- if the length is not equal then return false 
- if length is same then check for the no.of times the character is occuring , returns true if it satisfies these
- return flase if count doesnt match 

```python
def ana(s1,s2):
    s1 = s1.lower().replace(" ","")
    s2 = s2.lower().replace(" ","")
    if len(s1)!=len(s2):
        return False
    for char in s1:
        if s1.count(char)!=s2.count(char):
            return False
    return True
s1 = "listen"
s2 = "silent"
s = ana(s1,s2)
print(s)
print()
s1 = "rack"
s2 = "car"
s = ana(s1,s2)
print(s)
```

#6 
- create empty list 
- if numbers are not zeros store them in new list 
- find zeros by subtracting the length of new list by original list of numbers length
- add that many no.of zeros and return the result

```python

def zeros(li):
    w = []
    for l in li:
        if l!=0:
            w.append(li)
    count = len(li) - len(w)
    for i in range(count):
        w.append(0)
    return w
li = [0, 3, 0, 1, 2, 0, 4]
e = zeros(li)
print(e)
print()
li = [0,1,3,0,0,0,2]
q = zeros(li)
print(q)
```

#7 
- create a list to store the unique numbers
- if no unique , return none
- sort them in order and return the last second number


```python
def sh(n):
    u = []
    for i in n:
        if i not in u:
            u.append(i)
    if len(u)<2:
        return None
    u.sort()
    return u[-2]
n = [10, 20, 40, 30, 40, 20]
w = sh(n)
print(w)
print()
n = [23,11,568,-21,4,0]
d = sh(n)
print(d)
```

#9 
- first find the sum of the n numbers
- then subtract the given numbers sum from the first sum
- the diffference number is the output which is missing number from n numbers


```python
def missing(n,m):
    total = m*(m+1)//2
    sum = 0
    for i in n:
        sum = sum+i
    miss = total-sum
    return miss
n = [1,2,4,5,6]
m = 6
l = missing(n,m)
print(l)
print()
n = [1,2,3,4,5,6]
m = 7
b = missing(n,m)
print(b)
```

#10 
- assume list and target
- choose position as second position and go through each element
- for that position check all previous positions
- if you get sum as target number then return them in list 

```python
def find(n,target):
    for j in range(1,len(n)):
        for i in range(j):
            if n[i]+n[j] == target:
                return [n[i],n[j]]
    return []
n = [2, 7, 11, 15]
target = 9
d = find(n,target)
print(d)
print()
n = [1,2,3,4,0]
target = 2
x = find(n,target)
print(x)
```

#11
- if list is empty then return none
- assume first value as closest and compare it with all elements with absolute values
- choose the closer value

```python

def close(n):
    if len(n)==0:
        return None
    closest = n[0]
    for i in n:
        if abs(i)<abs(closest):
            closest = i
        elif abs(i) == abs(closest) and i>closest:
            closest=n
    return closest
n = [-8, -3, 2, 5, -2]
j = close(n)
print(j)
print()
n = [2,1,0,-1,8]
g = close(n)
print(g)
```

#14
- assume the maximum sum as sum of last two elements 
- go through each consecutive pairs
- calculate the current sum and compare with max sum 
- update the max sum if current sum is greater than that
return max sum

```python
def max(n):
    max_sum = n[0]+n[1]
    for i in range(1,len(n)-1):
        sum = n[i]+n[i+1]
        if sum > max_sum:
            max_sum = sum
    return max_sum
n = [4, 7, 2, 9, 5]
e = max(n)
print(e)
print()
n = [3,1,-2,0]
s = max(n)
print(s)
```

#15 
- assume a target value
- create empty list to store result
- for different positions create different loops 
- if sum is equal to target then return the pair in a sorted way
- add the pair if it not in reult list
return the list

```python
def pairs(n,target):
    r = []
    for i in range(len(n)):
        for j in range(i+1,len(n)):
            if n[i]+n[j]==target:
                pair = [n[i],n[j]]
                pair.sort()
                if pair not in r:
                    r.append(pair)
    r.sort()
    return r
n= [2, 4, 3, 5, 7, 8, 1]
target = 9
b = pairs(n,target)
print(b)
print()
n= [0,1,-2,9,6,3,4,7,22]
target = 7
k = pairs(n,target)
print(k)
```
